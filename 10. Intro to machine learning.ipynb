# Machine learning (ML)

_Machine learning (ML) is a branch of artificial intelligence (AI) focused on building algorithms that learn patterns directly from data to make predictions, decisions, or uncover insights without being explicitly programmed with hardcoded rules._<br>

Machine Learning builds upon the insights gained during Exploratory Data Analysis (EDA) by transforming data into action. While EDA focuses on understanding data distributions, identifying patterns, detecting anomalies, and assessing relationships among variables, Machine Learning uses these insights to develop algorithms that can learn from historical data and make predictions or classifications on new, unseen observations. 

In this section, we will introduce the fundamental concepts of Machine Learning, including the different types of learning, the model development process, model evaluation techniques, and the practical considerations involved in building reliable predictive systems.

## 1. ML Definition and uses

In traditional software development, programmers write explicit rules and logic to process inputs and produce outputs. In Machine Learning, the paradigm is inverted. You supply historical inputs along with the known historical outcomes, and the algorithm automatically extracts the underlying mathematical function connecting them. 

Machine learning is essential when traditional rule-based logic fails:

- *Problems with dynamic or evolving patterns* such as fraud patterns, spam messaging, and stock trends evolve continuously. Static code quickly breaks down, whereas ML models can adapt as new data arrives.

- *Too complex for human-crafted rules* like tasks like identifying objects in an image, understanding natural speech, or diagnosing diseases from clinical scans require analyzing thousands of high-dimensional variables simultaneously, or

- *Large-scale automation.* ML enables personalized recommendations for millions of users, automated customer support, and predictive maintenance at scale without manual human intervention.

## 2. Primary Types of Machine Learning

Machine learning problems are broadly categorized into four primary paradigms based on how data is structured and how feedback is provided to the learning algorithm. 

- __Unsupervised Learning__ deals with unlabeled datasets where only the input features (X) are provided without any target labels. Instead of making explicit predictions, the algorithm analyzes raw data to uncover hidden structures, natural distributions, or underlying patterns independently. Its primary applications include clustering, which groups similar data points together for tasks like customer segmentation, and dimensionality reduction, which compresses complex high-dimensional datasets into simpler representations while preserving key information.

    - The most common algorithms in this type are:        
        - Hierarchical Clustering, 
        - K-Means Clustering, 
        - Principal Component Analysis (PCA), (The components produced can be used as new variables in any regression model*)
        - DBSCAN, and 
        - Gaussian Mixture Models 
        
- __Supervised Learning__ is the most common paradigm in machine learning, relying on fully labeled datasets where every input instance (predictor variable) is paired with a known ground-truth target (predicted variable). The goal of the algorithm is to learn a mapping function that can accurately predict outcomes for brand-new, unseen data. Supervised learning splits into two main problem types: regression, which predicts continuous numerical outputs like house prices or temperature, and classification, which assigns discrete categories like identifying spam emails or diagnosing medical scans.

    - The most common algorithms for classification in this type are:        
        - Logistic Regression, 
        - k-Nearest Neighbors (k-NN), 
    
    - The most used algorithms for regression analysis are: 
        - Linear regression, 
        - polynomial regression, 
    
    - Some algorithms can be used for both regression and classification:
        - Decision Trees, 
        - Random Forests, 
        - Gradient Boosting Trees (XGBoost, LightGBM, CatBoost), 
        - Support Vector Machines (SVM), and 
        - Neural Networks

- __Semi-Supervised Learning__ bridges the gap between supervised and unsupervised approaches by training on a small amount of labeled data paired with a significantly larger volume of unlabeled data. Labeling vast amounts of data manually is often expensive, slow, or impractical—such as in specialized medical imaging or complex audio transcription. By utilizing the small labeled dataset to guide target mapping while drawing broader patterns and structural insights from the unlabeled pool, semi-supervised models achieve high accuracy at a fraction of the data annotation cost.

    - The most common algorithms in this type are:        
        - Self-Training, 
        - Label Propagation, 
        - Pseudo-Labeling, 
        - Semi-Supervised SVMs, 
        - Consistency Regularization Methods

- __Reinforcement Learning__ operates without a static historical dataset, taking an interactive approach to machine learning inspired by behavioral trial and error. An autonomous agent learns by making decisions within a dynamic environment, receiving positive rewards for favorable actions and negative penalties for mistakes. The objective is to discover an optimal policy (sequence of actions) that maximizes total cumulative reward over time. This approach underpins autonomous driving systems, complex game-playing algorithms, robotic control, and the human feedback alignment (RLHF) used in modern large language models.

    - The most common algorithms in this type are:        
        - Q-Learning, 
        - SARSA, 
        - Deep Q Networks (DQN), 
        - Policy Gradient Methods, 
        - Actor-Critic Methods, 
        - Proximal Policy Optimization (PPO)

## __Summary__

Machine Learning extends Exploratory Data Analysis by shifting software from fixed, human-coded rules to data-driven discovery, enabling systems to solve high-dimensional, dynamic problems at scale. In this section we explored the types of machine learning, its uses, and most common tools.

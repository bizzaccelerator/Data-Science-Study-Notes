# Numerical Python (Numpy)

NumPy (Numerical Python) is a powerful Python library built on C, acting as a wrapper around C functions to leverage its superior execution speed. Numpy improves Python language capabilities to operate with multidimensional arrays and matrices, along with a large collection of high-level mathematical functions to operate on these arrays.

Unlike standard Python lists—which are flexible arrays containing pointers to objects scattered in memory—NumPy arrays consist of fixed-type data (such as continuous blocks of 32-bit integers).

This architecture allows NumPy to process operations much faster than Python lists for three key reasons:

- Uniform Data Types: Because all elements share the exact same data type, operations can be executed directly without constant type-checking.

- Reduced Overhead: Standard Python list items must store size, reference counts, object types, and values individually. NumPy's streamlined structure eliminates this overhead, resulting in significantly faster processing and lower memory consumption.

- Contiguous Memory Allocation: NumPy stores its data in continuous blocks right next to each other in memory. In contrast, standard Python lists scatter items across memory, forcing the CPU to waste time jumping back and forth to retrieve them. Continuous memory allows for much faster data access and better utilization of modern CPU caching.

Therefore, we can conclude that *Python Lists* are designed for general-purpose programming and data organization. They support operations like appending, inserting, and removing items easily, but do not support native mathematical operations (e.g., adding two lists together concatenates them rather than adding their elements mathematically), while *NumPy Arrays* are designed specifically for mathematical and numerical computations. They natively support vectorized math, matrix multiplications, linear algebra, statistical operations, and broadcasting.

Other libraries like Tensorflow or Scipy are built on Numpy. It worth the time!

## 2. LINEAR ALGEBRA

## 2.1. Eigenvalues and Eigenvector

Eigenvalues and eigenvectors are special numbers and vectors associated with linear transformations, defined by the fundamental equation Av = λ v, where A is a matrix, v is the eigenvector, and λ is the eigenvalue.

- An __Eigenvector (v)__ is a non-zero vector that does not change direction when a linear transformation (matrix multiplication) is applied to it; it only changes by a scaling factor. 

- An __Eigenvalue (λ)__ is the scalar factor by which the corresponding eigenvector is scaled (stretched, compressed, or reversed) during the transformation.

Straight eigenvalues and eigenvectors decomposition can be applied to squared matrices only. For non-squared matrices, ither methods such as Singular Value Decomposition (SVD) should be used.

Their principal uses are simplifying powers of matrices and systems of differential equations (Matrix diagonalization), reducing data dimensions using tools like Principal Component Analysis (PCA), and pawering Algorithms such as *power ranking* in search engines like Google's PageRank. Know more [here](https://math.mit.edu/~gs/linearalgebra/ila5/linearalgebra5_6-1.pdf)

```python
# Take an squared 2d array of random integers between 0 and 20 with shape (4, 4)
s_2d_array = rg.integers(20, size=(4, 4)) 
s_2d_array
```

```python
#  The decomposition in eigenvalues and eigenvectors of a square matrix can be computed using the linalg.eig function as follows
eigenvalues, eigenvectors = np.linalg.eig(s_2d_array)

# The eigenvalues is a column vectorwith the eigenvalues of the matrix
print("Eigenvalues: \n", eigenvalues)
```

```python
# The eigenvectors is a matrix with the corresponding eigenvectors as columns. 
print("Eigenvectors: \n", eigenvectors)
 
```

## 2.2. Solving linear systems

A linear system of equations is a set of two or more linear equations sharing the same variables. These linear systems of equations may have infinitely, unique or no solution. A solution of a linear system is an assignment of values to the variables such that each of the equations is satisfied. The set of all possible solutions is called the solution set.

Lets say we have a the following system:

x1 + x2 = 2200 <br>
1.5x1 + 4x2 = 5500


```python
# This system can be rewrited as a matrix equation Ax = b, 
# where A is the matrix of coefficients, x is the vector of unknowns, and b is the vector of constants.

A = np.array([[1, 1], [1.5, 4]])
b = np.array([2200, 5050])
```

```python
# which solution is x = A^-1 * b. Manually is

manual_solution = np.linalg.inv(A).dot(b) # This calculates the inverse of A and multiplies it by b to get the solution vector x
print("The manual solution is:", manual_solution)

# The manual solution can lead to numerical instability and is not recommended for solving linear systems. 
# Instead, we can use the solve function as follows

prefered_solution = np.linalg.solve(A,b)
print("The preferred solution is:", prefered_solution)
```

```python
# We can compare the solutions obtained by both methods to see if they are the same.
manual_solution == prefered_solution
```

```python
# The comparison above leads to a misunderstanding of the fact that the two solutions are exactly the same. 
# The False conclusion obtained is due to numerical precision issues. Hence, a better way to compare them is 
np.allclose(manual_solution, prefered_solution)
```

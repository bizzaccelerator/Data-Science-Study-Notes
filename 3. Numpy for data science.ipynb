# Numerical Python (Numpy)

NumPy (Numerical Python) is a powerful Python library built on C, acting as a wrapper around C functions to leverage its superior execution speed. Numpy improves Python language capabilities to operate with multidimensional arrays and matrices, along with a large collection of high-level mathematical functions to operate on these arrays.

Unlike standard Python lists—which are flexible arrays containing pointers to objects scattered in memory—NumPy arrays consist of fixed-type data (such as continuous blocks of 32-bit integers).

This architecture allows NumPy to process operations much faster than Python lists for three key reasons:

- Uniform Data Types: Because all elements share the exact same data type, operations can be executed directly without constant type-checking.

- Reduced Overhead: Standard Python list items must store size, reference counts, object types, and values individually. NumPy's streamlined structure eliminates this overhead, resulting in significantly faster processing and lower memory consumption.

- Contiguous Memory Allocation: NumPy stores its data in continuous blocks right next to each other in memory. In contrast, standard Python lists scatter items across memory, forcing the CPU to waste time jumping back and forth to retrieve them. Continuous memory allows for much faster data access and better utilization of modern CPU caching.

Therefore, we can conclude that *Python Lists* are designed for general-purpose programming and data organization. They support operations like appending, inserting, and removing items easily, but do not support native mathematical operations (e.g., adding two lists together concatenates them rather than adding their elements mathematically), while *NumPy Arrays* are designed specifically for mathematical and numerical computations. They natively support vectorized math, matrix multiplications, linear algebra, statistical operations, and broadcasting.

## 1.1. Creating arrays.

## 1.4. Trigonometric functions

```python
# trigonometric functions can be applied to arrays as follows 
from numpy import sin, cos, pi

radians = np.linspace(0, 2*pi, 100) # 100 evenly spaced values from 0 to 2*pi
radians
```

```python
# sine and Cosine functions can be simply applied to the array as
sine = sin(radians)
cosine = cos(radians)

# Sine values are:
print("Sine values: ", sine)
```

## 1.5. Operations with arrays

Multiple operations can be performed in a simple way. 

```python
# Recall arr1 and arr2 are 1D arrays, we can perform element-wise operations on them as follows
print("Recall arr1 and arr2 are 1D arrays:")
print("Array 1: ", arr1)
print("Array 2: ", arr2)
```

```python
print("we can perform element-wise operations on them as follows:")
sum_arr = arr1 + arr2 # It adds ith element of arr1 to ith element of arr2
diff_arr = arr1 - arr2 # it subtracts ith element of arr2 from ith element of arr1
prod_arr = arr1 * arr2 # it performs the dot product of the two arrays, multiplying ith element of arr1 by ith element of arr2
quot_arr = arr1 / arr2 # it divides ith element of arr1 by ith element of arr2

print("Sum: ", sum_arr)
print("Difference: ", diff_arr)
print("Product: ", prod_arr)
print("Quotient: ", quot_arr)
```

Concatenation of arrays is highly used, it can be done as

```python
# Concatenation of arrays can be done using the concatenate function as follows
concatenated_arr = np.concatenate([arr1, arr2]) # This created a whole new array 
print("Concatenated Array: ", concatenated_arr)
```

```python
# organizing the elements of an array can be done using the sort function as follows
sorted_array = np.sort(concatenated_arr) # This sorts the array in ascending order
print("Sorted Concatenated Array: ", sorted_array)
```

## 1.6. Generating random numbers

Random values are important to validate hypothesis and many more.

```python
# Random numbers can be generated using the random module of numpy as follows
rg = np.random.default_rng(2) # Create a random number generator anchored in the seed of 2
# Using the same seed ensured obtaining the same random numbers each time the code is run, 
# which is useful for reproducibility in experiments and testing.
```

```python
# our random object rg is used to generate the random numbers as follows
random = rg.random(1000) # Create an array of 1000 random numbers between 0 and 1 using the archored seed of 2
random
```

```python
# Random data from a normal distribution can be generated using the normal function as follows
random_normal = rg.normal(10, 5, 100000) # 10000 random numbers with a mean of 10 and a standard deviation of 5
random_normal
```

```python
# An array of integers can be generated using the integers function as follows
integers = rg.integers(0, 100, 2000) # 2000 random integers between 0 and 100
integers
```

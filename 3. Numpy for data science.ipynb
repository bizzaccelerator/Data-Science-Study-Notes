# Numerical Python (Numpy)

NumPy (Numerical Python) is a powerful Python library built on C, acting as a wrapper around C functions to leverage its superior execution speed. Numpy improves Python language capabilities to operate with multidimensional arrays and matrices, along with a large collection of high-level mathematical functions to operate on these arrays.

Unlike standard Python lists—which are flexible arrays containing pointers to objects scattered in memory—NumPy arrays consist of fixed-type data (such as continuous blocks of 32-bit integers).

This architecture allows NumPy to process operations much faster than Python lists for three key reasons:

- Uniform Data Types: Because all elements share the exact same data type, operations can be executed directly without constant type-checking.

- Reduced Overhead: Standard Python list items must store size, reference counts, object types, and values individually. NumPy's streamlined structure eliminates this overhead, resulting in significantly faster processing and lower memory consumption.

- Contiguous Memory Allocation: NumPy stores its data in continuous blocks right next to each other in memory. In contrast, standard Python lists scatter items across memory, forcing the CPU to waste time jumping back and forth to retrieve them. Continuous memory allows for much faster data access and better utilization of modern CPU caching.

Therefore, we can conclude that *Python Lists* are designed for general-purpose programming and data organization. They support operations like appending, inserting, and removing items easily, but do not support native mathematical operations (e.g., adding two lists together concatenates them rather than adding their elements mathematically), while *NumPy Arrays* are designed specifically for mathematical and numerical computations. They natively support vectorized math, matrix multiplications, linear algebra, statistical operations, and broadcasting.

Other libraries like Tensorflow or Scipy are built on Numpy. It worth the time!

## 1.1. Creating arrays.
The arrays are the core elements of Numpy. The first step is loading Numpy:

```python
# First install Numpy as pip install numpy
import numpy as np
```

```python
# list are created as follows
list1 = [1, 2, 3, 4, 5]
```

```python
# We can create a numpy array from the list already created or directly introducing the values in the array as follows
arr1 =np.array(list1)
arr2 = np.array([30, 21, 8, 14, 5])
```

```python
#To retreive the array we can use the print function as follows
print(arr1)
print(arr2)
```

Please notice the difference between lists and arrays when presented 

```python
print("A list is: ", list1, "and a numpy array is: ", arr1)
```

An array may be or higher order (n-dimentional). Hence, These arrays can be created as:

```python
arr_2d = np.array([[24,14,7],[51,3,95]])

arr_2d 
```

```python
# We can create empty matrices with zeros or ones as follows
arr_zeros = np.zeros((3, 4)) # 3 rows and 4 columns
arr_zeros
```

```python
# A matix full of ones is useful too
arr_ones = np.ones((2, 3)) # 2 rows and 3 columns
arr_ones
```

```python
# A unidimensional array can be created with the arange function as follows
arr_range = np.arange(0, 60, 2) # from 0 to 59 with a step of 2
arr_range
```

```python
# A controled array can be created with the linspace function as follows
arr_linspace = np.linspace(0, 5, 10) # A unidimensional array from 0 to 5 with 10 evenly spaced values
arr_linspace
```

The dimensions indicates the number of axes or the rank of the array. In the precedent example, 2 dimensions.

## 1.4. Trigonometric functions

```python
# trigonometric functions can be applied to arrays as follows 
from numpy import sin, cos, pi

radians = np.linspace(0, 2*pi, 100) # 100 evenly spaced values from 0 to 2*pi
radians
```

```python
# sine and Cosine functions can be simply applied to the array as
sine = sin(radians)
cosine = cos(radians)

# Sine values are:
print("Sine values: ", sine)
```

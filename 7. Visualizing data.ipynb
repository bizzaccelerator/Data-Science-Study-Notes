# What is Data Visualization?

Data visualization is the process of putting numerical data into a visual format—like charts, graphs, maps, or dashboards—so people can see and understand what the data means at a glance. In simple terms: It turns raw numbers into a picture.

## 1. Why is Data Visualization Important?

- Speeds up understanding: Humans process visual information far faster than rows of spreadsheet data. A simple line chart immediately shows a trend that might take minutes to find in a table.

- Spot patterns and outliers easily: Visuals make it instant to notice hidden trends, repeating patterns, or unusual data points (outliers) that would otherwise blend into thousands of rows.

- Simplifies complex ideas: It translates heavy statistical analysis or machine learning results into plain, easy-to-understand stories for non-technical teammates, managers, or clients.

- Drives better decisions: When data is presented clearly and visually, decision-makers can draw accurate conclusions and act on insights much faster.


## 2. How to spot insights?
 
Using images! There are different graphs useful to illustrate relationship between variables. Here below the most frequently used illustrations:

![chart_types.png](./Images/chart_types.png)

### Line Charts
- __What they show:__ Changes over a continuous period (time).
- When to use: Tracking trends over time, such as stock prices, monthly sales, or daily temperature changes.

### Bar Charts
- __What they show:__ Comparisons between discrete categories.
- When to use: Comparing quantities across groups, such as revenue by product category or population by city.

### Scatter Plots
- __What they show:__ The relationship between two continuous numerical variables.
- When to use: Checking if two variables are correlated, such as comparing study hours versus exam scores.

### Histograms
- __What they show:__ How continuous numerical data is distributed across ranges (bins).
- When to use: Seeing how values are spread out, such as the age distribution of website visitors or salary ranges in a company.

### Box Plots (Box-and-Whisker)
- __What they show:__ A summary of data distribution based on minimum, first quartile, median, third quartile, and maximum values.
- When to use: Comparing distributions across multiple groups and instantly spotting outliers.

### Heatmaps
- __What they show:__ Data values as a color-coded matrix.
- When to use: Displaying correlation matrices between variables or showing activity patterns (like peak website traffic hours by day of the week).

## 3. What tools are available?

When you start exploring data, you quickly realize that different libraries serve different purposes depending on what you're trying to build. Matplotlib, Seaborn, and Plotly do not behave the same. 


- __1. Matplotlib__ is relevant because it is the backbone of Python data visualization. Almost every python data science library relies on it. It is best for complete custom control over every tiny detail (axes, ticks, colors, labels), which implies it is Highly customizable, but requires a lot of code. It is recommended when you need standard basic charts (line, scatter, bar) or fine-grained control over a plot's layout.

- __2. Seaborn__ is relevant because it makes complex statistics easy and attractive. It automates the tedious code Matplotlib requires for standard data exploration. It shines when you want clean, beautiful plots (like heatmaps, box plots, or violin plots) out of the box with minimal code.

- __3. Plotly__ is relevant because it brings charts to life with interactivity. It bridges the gap between static analysis and interactive web dashboards or user reports. Its best use case emerge when dashboards, web reports, and graphs you can zoom into, hover over, or click. Plotly is recommended when static images aren't enough and you need users to interact with the data directly in a browser or notebook.

Matplotlib & Seaborn are connected directly because seaborn actually uses Matplotlib behind the scenes. You can combine them in the same script—using Seaborn to generate a quick, beautiful chart, and Matplotlib commands to fine-tune the titles or axes. However, Plotly operates independently. Plotly uses JavaScript under the hood, producing interactive HTML elements you can manipulate.

## 4. Matplotlib (The basics)

The graphs with matplotlib are built always using a standar architecture. The key elements are:

- __Figure:__ The top level container (The entire window/page)
- __Axes:__ The actual plot area where the data is drawn.
- __Axis:__ The number line objects (x-axis, y-axis)

Think of Figure as a canvas and Axes as individual paintings on that canvas.


__WARNING: It never works alone!__. It illustrate the dta you are working with in pandas, numpy, or any other library for data processing. Hence, they are imported together.

```python
# First install Matplotlib as pip install matplotlib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# For Jupyternotebook, we use this magic command to display the plots inline
%matplotlib inline

# Optional: Set style for better aesthetics
plt.style.use('default') # OTher options include 'ggplot', 'seaborn', etc.
```

### 4.1. Line plot

A line plot connects individual data points with a line. It is used to show how a continuous variable changes over time or across a sequential order.

```python
# the first Method is directly illustrating it

# First start with the data of interest. Here we will use a simple example of monthly sales data using lists. 
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [200, 250, 300, 400, 350]

# then plot the line
plt.plot(months, sales) # using months as x-axis and sales as y-axis

# Add title and labels
plt.title('Monthly Sales Data')
plt.xlabel('Month')
plt.ylabel('Sales') 
plt.grid(True) # Optional: Add grid for better readability

# Display the graph
plt.show() # Not necessary to use plt.show() in Jupyter Notebook, but it is mandatory to include it for scripts in order to display the plot when running the script outside of Jupyter Notebook.
```

There is a second method: the Object oriented approach, which offers more control. It is recommended for complex plots. This method allow you to chose what and hot to show the information. 


```python
# Using the subplots() function to create multiple plots in a single figure.
fig, ax = plt.subplots() # Create a figure and an axes. 
ax.plot(
    months, # X axis data
    sales,  # Y axis data
    marker='o', # Marker style for data points
    linestyle='-', # Line style for the plot
    color='b', # The color of the line
    label='Sales' # Label for the legend
    )
ax.set_title('Monthly Sales Data') # Set the title of the plot
ax.set_xlabel('Month') # Set the x-axis label
ax.set_ylabel('Sales') # Set the y-axis label
ax.grid(True) # Optional: Add grid for better readability

plt.show() # Display the graph
```

There are many lines we can draw as below:

```python
# First, it's always the data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)

# Creating the graph, this is the way to include multilines in single images
plt.figure(figsize=(8, 5)) # Set the size of the figure (width, height) in inches
plt.plot(x, y1, label='Sin(x) values') 
plt.plot(x, y2, label='Cos(x) values')
plt.plot(x, y3, label='Sin(x)*Cos(x) values')

# configuring the illustration
plt.xlabel('X values')
plt.ylabel('Trigonometric functions')
plt.title('Multiple functions')
plt.grid(True, alpha=0.3) # Alpha controls the transparency 
plt.legend() # Show the legend

# show
plt.show()
```

### There are many way of customizing the graphs created. Some options are

```python
# First the data
x = np.linspace(1, 10, 100)
y = x*x

# Then create the figure
plt.Figure(figsize=(15,8))
plt.plot(
    x, # The data in x
    x, # The data in y
    'r-', # A short way to indicate "color red" and "solid line"
    label = 'Solid line', # Description of the data 
    linewidth = 4 # The width
)
plt.plot(x, y, 'g--', label= 'dashed line', linewidth= 6)

# Configure the graph
plt.title('Line customized')
plt.xlabel('Dummy data')
plt.ylabel('Dummy mirror')
plt.legend()
plt.grid(True)


# Show the graph
plt.show()
```

### 4.2. Bar plot

A bar plot (or bar chart) uses rectangular bars to show comparisons between distinct groups or categories. The length or height of each bar is proportional to the value it represents.

__Vertical Bar Charts__

```python
# The data is 
categories = ['A', 'B', 'C', 'D']
values = [100, 125, 80, 92]

# Create the graph
plt.Figure(figsize=(8,6))
plt.bar( # Bar is the command for bar plots using categories in x and values in y
    categories,
    values,
    color = 'steelblue',
    edgecolor = 'black', 
    linewidth = 1.2
)

# Configure the image
plt.title('Bar plot')
plt.xlabel('Categories of interes')
plt.ylabel('Frequency of categories')
plt.grid(True, alpha = 0.5)

# Show the graph
plt.show()
```

__Horizontal Bar Charts__ are useful too. They operates with the hbar() command


```python
# The data is 
categories = ['A', 'B', 'C', 'D']
values = [100, 125, 80, 92]

# Create the figure
plt.Figure(figsize=(8, 6))
plt.barh(
    categories,
    values,
    color = 'coral',
    edgecolor = 'magenta',
    linewidth = 1.2
)

# Configure the image
plt.title('Horizontal Bar plot')
plt.xlabel('Frequency of categories')
plt.ylabel('Categories of interes')
plt.grid(True, alpha = 0.5)

# Show the graph
plt.show()
```

Multiple bars can be combined together easily with matplotlib. 

```python
# Grouped bar chart
categories = ['Q1', 'Q2', 'Q3', 'Q4']
product_a = [200, 220, 400, 250]
product_b = [250, 220, 380, 200]
product_c = [270, 250, 390, 240]

x = np.arange(len(categories)) # Creates a numerical vector for x axis
width = 0.25 # indicates the whith of the bars

# Create the graph
plt.Figure(figsize=(12,8))
plt.bar(x - width, product_a, width, label = 'Product A', color = '#FF6B6B')
plt.bar(x, product_b, width, label = 'Product B', color = '#4ECDC4')
plt.bar(x + width, product_c, width, label = 'Product C', color = '#95E1D3')

# Configure the graph
plt.title('Sales of portfolio of product')
plt.xlabel('Quarter')
plt.ylabel('Sales')
plt.xticks(x, categories) # Asociate the x positions to the categories in X
plt.grid(True, alpha = 0.3)
plt.legend()

# Show the graph
plt.show()
```

### 4.3. Scatter plot

A scatter plot uses individual dots plotted on a two-dimensional grid to show the relationship (or correlation) between two continuous numerical variables.

```python
# the data is
np.random.seed(42) # Recall from numpy section this ensures reproducibility 
x = np.random.randn(100)
y = 2 * x + np.random.randn(100)*0.5 # A linear regression with noise

# Create the figure
plt.Figure(figsize= (8,6))
plt.scatter(
    x,
    y, 
    alpha= 0.6 # Controls the transparency. Where 1 is fully solid
)

# Configure the image
plt.title('Scatter plot')
plt.xlabel(' X values')
plt.ylabel('Y values')
plt.grid(True, alpha=0.3)

# Show the data
plt.show()
```

Scatter plots can be used for more prolific graphs

```python
# The data
np.random.seed(42)
x = np.random.randn(100)
y = np.random.randn(100)

colors = np.random.randn(100) # Create random number for color enconding
# 100 is used to ensure the three series (x, y, and colors) are of the same lenght. 

sizes = 1000 * np.random.randn(100) # Create a series of random sizes for each combination (x,y). 

# Create the graph
plt.Figure(figsize=(12,8))
plt.scatter(
    x, 
    y,
    c=colors, # Setting the color vectors
    s = sizes, # Using the size series created
    cmap='viridis' # Color pallet used
)
plt.colorbar(label='Color intensity') # Add the colorbar for explaining the intensity

# configure the illustration
plt.title('Scatter plot with color and size variations')
plt.xlabel('X values')
plt.ylabel('Y values')
plt.grid(True, alpha = 0.3)

# Show it
plt.show()
```

### 4.4. Histogram plot

A histogram looks similar to a bar chart, but it serves a very different purpose. It groups continuous numerical data into ranges called bins to show the distribution (shape, spread, and central tendency) of a single dataset. Therefore, the main difference between a histogram and a bar graph is the type of data they show: bar graphs compare separate categories, while histograms show how numerical data spreads out over a continuous range. __Histograms comes in hand to verify the distribution of data.__


```python
# Always the data comes first
np.random.seed(42)
data1 = np.random.normal(100, 15, 1000)
data2 = np.random.normal(120, 20, 1000)

# Create the figure
plt.Figure(figsize=(15,8))
plt.hist(
    data1, # The series to draw
    bins=30, # The nomber of bins to use
    alpha=0.6, 
    label='Dataset 1',
    color='blue'
)
plt.hist(data2, bins=30, alpha = 0.6, label='Dataset 2', color = 'red')

# Configure the data
plt.title('Overlaping histograms')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.legend()
plt.grid(True, alpha = 0.3, axis = 'y')

# Show
plt.show()
```
